```python

import duckdb
from pathlib import Path

# Create DuckDB connection
con = duckdb.connect()

# Function to read CSV safely
def get_table(path):
    return f"""
        read_csv(
            '{path.as_posix()}',
            header = true,
            all_varchar = true,
            ignore_errors = false
        )
    """

# Audit Email, File and HTTP datasets
for name in ["Email", "File", "HTTP"]:

    path = datasets[name]
    table = get_table(path)

    print("\n" + "=" * 60)
    print(f"DATA QUALITY AUDIT: {name}")
    print("=" * 60)

    # 1. Total records and unique IDs
    total, unique = con.execute(f"""
        SELECT
            COUNT(*),
            COUNT(DISTINCT id)
        FROM {table}
    """).fetchone()

    print("Total records:", f"{total:,}")
    print("Unique IDs:", f"{unique:,}")
    print("Duplicate IDs:", f"{total - unique:,}")

    # 2. Missing values in important columns
    print("\nMissing values:")

    for col in ["id", "date", "user", "pc"]:

        missing_query = f"""
            SELECT COUNT(*)
            FROM {table}
            WHERE "{col}" IS NULL
               OR TRIM("{col}") = ''
        """

        missing = con.execute(missing_query).fetchone()[0]

        print(f"{col}: {missing:,}")

    # 3. Date range
    print("\nDate range:")

    min_date, max_date = con.execute(f"""
        SELECT
            MIN(TRY_STRPTIME(date, '%m/%d/%Y %H:%M:%S')),
            MAX(TRY_STRPTIME(date, '%m/%d/%Y %H:%M:%S'))
        FROM {table}
    """).fetchone()

    print("Start:", min_date)
    print("End:", max_date)

print("\nDATA QUALITY AUDIT COMPLETED")
```

```python

for name, path in datasets.items():

    table = get_table(path)

    print("\n" + "=" * 60)
    print(f"COMPLETE MISSING VALUE AUDIT: {name}")
    print("=" * 60)

    # Get column names
    columns = con.execute(
        f"DESCRIBE SELECT * FROM {table}"
    ).fetchall()

    # Build one query to scan all columns together
    expressions = []

    for column in columns:
        col = column[0]

        expressions.append(f"""
            COUNT(*) FILTER (
                WHERE "{col}" IS NULL
                   OR TRIM("{col}") = ''
            ) AS "{col}"
        """)

    query = f"""
        SELECT {", ".join(expressions)}
        FROM {table}
    """

    result = con.execute(query).fetchone()

    found = False

    for column, missing in zip(columns, result):
        if missing > 0:
            found = True
            print(f"{column[0]}: {missing:,}")

    if not found:
        print("No missing values found in any column.")

print("\nCOMPLETE MISSING VALUE AUDIT FINISHED")
```

```python

# Compare employee IDs across consecutive LDAP snapshots

monthly_users = {}

for file in sorted(LDAP_DIR.glob("*.csv")):
    df = pd.read_csv(file)
    monthly_users[file.stem] = set(df["user_id"].dropna())

months = sorted(monthly_users.keys())

changes = []

for i in range(1, len(months)):

    previous_month = months[i - 1]
    current_month = months[i]

    previous_users = monthly_users[previous_month]
    current_users = monthly_users[current_month]

    added = current_users - previous_users
    removed = previous_users - current_users

    changes.append({
        "month": current_month,
        "new_employees": len(added),
        "employees_disappeared": len(removed),
        "previous_count": len(previous_users),
        "current_count": len(current_users)
    })

changes_df = pd.DataFrame(changes)

display(changes_df)

print("\nTotal new employee appearances:",
      changes_df["new_employees"].sum())

print("Total employee disappearances:",
      changes_df["employees_disappeared"].sum())
```

```python
logon_daily_query = f"""
WITH parsed AS (
    SELECT
        user,
        TRY_STRPTIME(date, '%m/%d/%Y %H:%M:%S') AS event_time,
        LOWER(activity) AS activity
    FROM {get_table(datasets["Logon"])}
)

SELECT
    CAST(event_time AS DATE) AS activity_date,
    user,
    COUNT(*) FILTER (
        WHERE activity = 'logon'
    ) AS logon_count,

    COUNT(*) FILTER (
        WHERE activity = 'logoff'
    ) AS logoff_count,

    COUNT(*) FILTER (
        WHERE EXTRACT(HOUR FROM event_time) < 8
           OR EXTRACT(HOUR FROM event_time) >= 18
    ) AS after_hours_event_count

FROM parsed
GROUP BY activity_date, user
ORDER BY activity_date, user
"""

logon_daily_df = con.execute(logon_daily_query).df()

display(logon_daily_df.head(10))
```

```python

email_daily_query = f"""
WITH parsed AS (
    SELECT
        user,
        TRY_STRPTIME(date, '%m/%d/%Y %H:%M:%S') AS event_time,
        "to" AS recipients,
        TRY_CAST(size AS BIGINT) AS email_size,
        TRY_CAST(attachments AS INTEGER) AS attachment_count
    FROM {get_table(datasets["Email"])}
)

SELECT
    CAST(event_time AS DATE) AS activity_date,
    user,

    COUNT(*) AS email_count,

    SUM(
        COALESCE(attachment_count, 0)
    ) AS total_attachment_count,

    COUNT(*) FILTER (
        WHERE COALESCE(attachment_count, 0) > 0
    ) AS emails_with_attachments,

    SUM(COALESCE(email_size, 0)) AS total_email_size,

    COUNT(*) FILTER (
        WHERE EXTRACT(HOUR FROM event_time) < 8
           OR EXTRACT(HOUR FROM event_time) >= 18
    ) AS after_hours_email_count

FROM parsed
GROUP BY activity_date, user
ORDER BY activity_date, user
"""

email_daily_df = con.execute(email_daily_query).df()

display(email_daily_df.head(10))
```

```python

http_daily_query = f"""
WITH parsed AS (
    SELECT
        user,
        TRY_STRPTIME(date, '%m/%d/%Y %H:%M:%S') AS event_time,
        url
    FROM {get_table(datasets["HTTP"])}
)

SELECT
    CAST(event_time AS DATE) AS activity_date,
    user,

    COUNT(*) AS http_count,

    COUNT(DISTINCT url) AS unique_urls,

    COUNT(*) FILTER (
        WHERE EXTRACT(HOUR FROM event_time) < 8
           OR EXTRACT(HOUR FROM event_time) >= 18
    ) AS after_hours_http_count

FROM parsed
GROUP BY activity_date, user
ORDER BY activity_date, user
"""

http_daily_df = con.execute(http_daily_query).df()

display(http_daily_df.head(10))
```

```python

file_daily_query = f"""
WITH parsed AS (
    SELECT
        user,
        TRY_STRPTIME(date, '%m/%d/%Y %H:%M:%S') AS event_time,
        filename
    FROM {get_table(datasets["File"])}
)

SELECT
    CAST(event_time AS DATE) AS activity_date,
    user,

    COUNT(*) AS file_copy_count,

    COUNT(DISTINCT filename) AS unique_files_copied,

    COUNT(*) FILTER (
        WHERE EXTRACT(HOUR FROM event_time) < 8
           OR EXTRACT(HOUR FROM event_time) >= 18
    ) AS after_hours_file_copy_count

FROM parsed
GROUP BY activity_date, user
ORDER BY activity_date, user
"""

file_daily_df = con.execute(file_daily_query).df()

display(file_daily_df.head(10))
```

```python
daily_dfs = {
    "Logon": logon_daily_df,
    "Device": device_daily_df,
    "Email": email_daily_df,
    "HTTP": http_daily_df,
    "File": file_daily_df
}

for name, df in daily_dfs.items():
    print(
        name,
        "| Rows:", len(df),
        "| Employees:", df["user"].nunique(),
        "| Duplicate keys:",
        df.duplicated(["activity_date", "user"]).sum()
    )
```

```python

import pandas as pd

# Step 1: Create a master list of all observed employee-date combinations
key_dfs = [
    df[["activity_date", "user"]]
    for df in daily_dfs.values()
]

master_keys = (
    pd.concat(key_dfs, ignore_index=True)
    .drop_duplicates()
    .sort_values(["activity_date", "user"])
    .reset_index(drop=True)
)

# Step 2: Merge all five daily feature DataFrames
master_df = master_keys.copy()

for name, df in daily_dfs.items():

    feature_columns = [
        col for col in df.columns
        if col not in ["activity_date", "user"]
    ]

    master_df = master_df.merge(
        df[["activity_date", "user"] + feature_columns],
        on=["activity_date", "user"],
        how="left",
        validate="one_to_one"
    )

# Step 3: Replace missing activity counts with zero
feature_columns = [
    col for col in master_df.columns
    if col not in ["activity_date", "user"]
]

master_df[feature_columns] = (
    master_df[feature_columns].fillna(0)
)

# Step 4: Sort the final table
master_df = master_df.sort_values(
    ["activity_date", "user"]
).reset_index(drop=True)

# Step 5: Display result
print("Master feature table shape:", master_df.shape)
print("Unique employees:", master_df["user"].nunique())
print("Duplicate employee-date keys:",
      master_df.duplicated(["activity_date", "user"]).sum())
print("Missing values:", master_df.isnull().sum().sum())

display(master_df.head(10))
```

```python

# Activity columns (excluding date and employee ID)
activity_columns = [
    col for col in master_df.columns
    if col not in ["activity_date", "user"]
]

# Employee-wise behavioural summary
employee_profile = (
    master_df
    .groupby("user")[activity_columns]
    .agg(["mean", "median", "std", "max"])
    .round(2)
)

print("Employee profile shape:", employee_profile.shape)

display(employee_profile.head(10))
```

```python

import pandas as pd

# Find each employee's first and last recorded activity
employee_periods = (
    master_df.groupby("user")["activity_date"]
    .agg(first_date="min", last_date="max")
    .reset_index()
)

# Generate a continuous daily calendar for each employee
calendar_parts = []

for row in employee_periods.itertuples(index=False):
    dates = pd.date_range(
        start=row.first_date,
        end=row.last_date,
        freq="D"
    )

    calendar_parts.append(
        pd.DataFrame({
            "activity_date": dates,
            "user": row.user
        })
    )

employee_calendar = pd.concat(
    calendar_parts,
    ignore_index=True
)

# Merge with existing activity features
complete_df = employee_calendar.merge(
    master_df,
    on=["activity_date", "user"],
    how="left",
    validate="one_to_one"
)

# Fill missing activity counts with zero
activity_columns = [
    col for col in complete_df.columns
    if col not in ["activity_date", "user"]
]

complete_df[activity_columns] = (
    complete_df[activity_columns].fillna(0)
)

complete_df = complete_df.sort_values(
    ["user", "activity_date"]
).reset_index(drop=True)

# Validation
print("Shape:", complete_df.shape)
print("Employees:", complete_df["user"].nunique())
print("Duplicate keys:",
      complete_df.duplicated(["activity_date", "user"]).sum())
print("Missing values:", complete_df.isnull().sum().sum())
print("Zero activity days:",
      (complete_df[activity_columns].sum(axis=1) == 0).sum())

display(complete_df.head(10))
```

```python

# Ensure correct sorting
complete_df = complete_df.sort_values(
    ["user", "activity_date"]
).reset_index(drop=True)

# Activity columns
activity_columns = [
    col for col in complete_df.columns
    if col not in ["activity_date", "user"]
]

# Calculate previous 7-day average for each employee
for col in activity_columns:

    complete_df[f"{col}_prev7d_mean"] = (
        complete_df.groupby("user")[col]
        .transform(
            lambda s: s.shift(1).rolling(
                window=7,
                min_periods=3
            ).mean()
        )
    )

# Validate
baseline_columns = [
    f"{col}_prev7d_mean"
    for col in activity_columns
]

print("Updated shape:", complete_df.shape)
print("7-day baseline features:", len(baseline_columns))

print(
    "Missing baseline values:",
    complete_df[baseline_columns].isnull().sum().sum()
)

display(
    complete_df[
        ["activity_date", "user", "http_count",
         "http_count_prev7d_mean",
         "device_connect_count",
         "device_connect_count_prev7d_mean"]
    ].head(15)
)
```

```python

# Calculate previous 30-day average for each activity feature

for col in activity_columns:

    complete_df[f"{col}_prev30d_mean"] = (
        complete_df.groupby("user")[col]
        .transform(
            lambda s: s.shift(1).rolling(
                window=30,
                min_periods=10
            ).mean()
        )
    )

# Validate
baseline_30d_columns = [
    f"{col}_prev30d_mean"
    for col in activity_columns
]

print("Updated shape:", complete_df.shape)

print(
    "30-day baseline features:",
    len(baseline_30d_columns)
)

print(
    "Missing 30-day baseline values:",
    complete_df[baseline_30d_columns].isnull().sum().sum()
)

display(
    complete_df[
        [
            "activity_date",
            "user",
            "http_count",
            "http_count_prev7d_mean",
            "http_count_prev30d_mean",
            "device_connect_count",
            "device_connect_count_prev7d_mean",
            "device_connect_count_prev30d_mean"
        ]
    ].head(15)
)
```

```python

# Create deviation features using 7-day historical averages

for col in activity_columns:

    baseline_col = f"{col}_prev7d_mean"

    deviation_col = f"{col}_deviation_7d"

    complete_df[deviation_col] = (
        complete_df[col] - complete_df[baseline_col]
    )

print("Dataset shape:", complete_df.shape)

deviation_columns = [
    f"{col}_deviation_7d"
    for col in activity_columns
]

display(
    complete_df[
        [
            "activity_date",
            "user",
            "http_count",
            "http_count_prev7d_mean",
            "http_count_deviation_7d",
            "device_connect_count",
            "device_connect_count_prev7d_mean",
            "device_connect_count_deviation_7d"
        ]
    ].head(15)
)
```

```python

# Extract day of week (Monday=0, Sunday=6)
complete_df["day_of_week"] = (
    complete_df["activity_date"].dt.dayofweek
)

# Weekend indicator
complete_df["is_weekend"] = (
    complete_df["day_of_week"] >= 5
).astype(int)

# Calculate total activity per day
complete_df["total_daily_activity"] = (
    complete_df[activity_columns].sum(axis=1)
)

# Activity performed during weekends
complete_df["weekend_activity"] = (
    complete_df["total_daily_activity"]
    * complete_df["is_weekend"]
)

# Validate
print("Weekend days:", complete_df["is_weekend"].sum())
print("Weekend activity events:",
      complete_df["weekend_activity"].sum())

display(
    complete_df[
        [
            "activity_date",
            "user",
            "day_of_week",
            "is_weekend",
            "total_daily_activity",
            "weekend_activity"
        ]
    ].head(15)
)
```

```python

# Aggregate after-hours activities across all five sources

after_hours_columns = [
    "after_hours_event_count",
    "after_hours_device_connect_count",
    "after_hours_email_count",
    "after_hours_http_count",
    "after_hours_file_copy_count"
]

# Total after-hours events per employee per day
complete_df["total_after_hours_activity"] = (
    complete_df[after_hours_columns].sum(axis=1)
)

# Proportion of activity occurring after hours
complete_df["after_hours_activity_ratio"] = (
    complete_df["total_after_hours_activity"]
    / complete_df["total_daily_activity"].replace(0, float("nan"))
)

# Weekend and after-hours combined activity
complete_df["weekend_after_hours_activity"] = (
    complete_df["total_after_hours_activity"]
    * complete_df["is_weekend"]
)

print("New after-hours features created.")

display(
    complete_df[
        [
            "activity_date",
            "user",
            "total_daily_activity",
            "total_after_hours_activity",
            "after_hours_activity_ratio",
            "weekend_after_hours_activity"
        ]
    ].head(15)
)
```

```python

import pandas as pd
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score
)

# Copy model predictions
evaluation_df = final_results_df.copy()

# Standardize employee IDs and dates
evaluation_df["user"] = evaluation_df["user"].astype(str)
evaluation_df["activity_date"] = pd.to_datetime(
    evaluation_df["activity_date"]
).dt.normalize()

# Initially label all employee-days as 0
evaluation_df["ground_truth"] = 0

# Mark documented incident days as 1
for _, incident in overlapping_incidents.iterrows():

    mask = (
        (evaluation_df["user"] == str(incident["user"])) &
        (evaluation_df["activity_date"] >= incident["start_date"]) &
        (evaluation_df["activity_date"] <= incident["end_date"])
    )

    evaluation_df.loc[mask, "ground_truth"] = 1

# Convert model predictions into binary labels
# -1 = Anomaly, +1 = Normal
evaluation_df["predicted_anomaly"] = (
    evaluation_df["anomaly_prediction"] == -1
).astype(int)

# Display ground-truth distribution
print("GROUND-TRUTH LABEL DISTRIBUTION:")
print(evaluation_df["ground_truth"].value_counts())

print("\nPREDICTION DISTRIBUTION:")
print(evaluation_df["predicted_anomaly"].value_counts())

# Calculate evaluation metrics
y_true = evaluation_df["ground_truth"]
y_pred = evaluation_df["predicted_anomaly"]

print("\nCONFUSION MATRIX:")
print(confusion_matrix(y_true, y_pred, labels=[0, 1]))

print("\nCLASSIFICATION REPORT:")
print(
    classification_report(
        y_true,
        y_pred,
        labels=[0, 1],
        target_names=["Normal/Undocumented", "Known Insider Incident"],
        zero_division=0
    )
)

print("\nKEY MODEL METRICS:")
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
```
